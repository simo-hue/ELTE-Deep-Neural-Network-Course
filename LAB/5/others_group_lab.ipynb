### Exercise 5.1: Min-max scaling

#### Problem Description

Solve the following tasks related to min-max scaling!

#### Task 1.

Define the `minmax_rescale(arr, a, b)` function, which linearly rescales the array `arr` into the interval between `a` and `b` such that the smallest element becomes `a` and the largest element becomes `b`.

Example:
```
arr = np.array([[2., 1.], [-2., 0.]])
minmax_rescale(arr, -1., 1.) -> ndarray([[1., 0.5], [-1., 0.]])
```

````python
# Implement your solution below
def minmax_rescale(vettore, minimo_interval, massimo_interval):
    real_min = np.min(vettore)
    real_max = np.max(vettore)
    
    res = []
    
    for elem in vettore:
        res.append(minimo_interval + (((elem - real_min) * (massimo_interval - minimo_interval)) / (real_max - real_min)))

    return res
    
# Test case
arr = np.array([[0.5, 1.2], [2.2, 1.0], [1.3, 0.9]])
arr1 = minmax_rescale(arr, -2., 3.)

assert np.allclose(arr1, np.array([[-2., 0.0588], [3., -0.529], [0.352, -0.823]]), atol=0.01)
print("Exercise 5.1, Task 1 passed! ")
````

#### Task 2.

Define the `minmax_rescale_per_var(arr, a, b)` function, which independently rescales the slices of the 2D array `arr` taken along axis #1 (the second axis) into the interval between `a` and `b`, such that within each slice the minimum becomes `a` and the maximum becomes `b`.
That is, for the result array `r`, every slice `r[i, :]` (i = 0, 1, ...) will have minimum `a` and maximum `b`.

Example:
```
arr = np.array([[2., 1., 0.], [-2., 0., 0.5]])
minmax_rescale_per_var(arr, -1., 1.) -> ndarray([[1., 0., -1.], [-1., 0.6, 1.]])
```

````python
# Implement your solution below
def minmax_rescale_per_var(arr, a, b):
    res = []
    for i in range(np.shape(arr)[0]):
        resultant_row = []
        
        real_min = np.min(arr[i,:])
        real_max = np.max(arr[i,:])
        
        for j in range(np.shape(arr)[1]):
            resultant_row.append(a + (((arr[i][j] - real_min) * (b - a)) / (real_max - real_min)))

        res.append(resultant_row)

    print(res)        
        
    return res

def minmax_rescale_per_var_pythonic(arr, a, b):
    mn = arr.min(axis=1, keepdims=True)   # shape (2, 1)
    mx = arr.max(axis=1, keepdims=True)
    return a + (arr - mn) * (b - a) / (mx - mn)

# Test case
arr = np.array([[2., 1., 0.], [-2., 0., 0.5]])
arr1 = minmax_rescale_per_var(arr, -1., 1.)
assert np.allclose(arr1, np.array([[1., 0., -1.], [-1., .6, 1.]]), atol=0.01)
print("Exercise 5.1, Task 2 passed! ")
````

#### Task 3.

Define the `minmax_rescale_along_axis(arr, a, b, axis)` function, which independently rescales the slices of the 2D array `arr` taken along the axis specified by `axis` into the interval between `a` and `b`, such that within each slice the minimum becomes `a` and the maximum becomes `b`.
That is, this should match `minmax_rescale_per_var` when `axis=1`. When `axis=0`, rescale the slices taken along the other axis. Try to handle both `axis=0` and `axis=1` in a single vectorized implementation, not with an `if` branch.

**Hint:** Check out the `keepdims` parameter of collapse-type functions.

Example:
```
arr = np.array([[2., 1., 0.], [-2., 0., 0.5]])
minmax_rescale_along_axis(arr, -1., 1., axis=0) -> ndarray([[1., 1., -1.], [-1., -1., 1.]])
minmax_rescale_along_axis(arr, -1., 1., axis=1) -> ndarray([[1., 0., -1.], [-1., 0.6, 1.]])
```

````python
# Implement your solution below
def minmax_rescale_along_axis(arr, a, b, axis):
    real_min = np.min(arr, axis=axis, keepdims=True)
    real_max = np.max(arr, axis=axis, keepdims=True)
    
    return a + (arr - real_min) * (b - a) / (real_max - real_min)
    
    
# Test case
arr = np.array([[2., 1., 0.], [-2., 0., 0.5]])
arr1 = minmax_rescale_along_axis(arr, -1., 1., axis=0)
assert np.allclose(arr1, np.array([[1., 1., -1.], [-1., -1., 1.]]), atol=0.01)
arr1 = minmax_rescale_along_axis(arr, -1., 1., axis=1)
assert np.allclose(arr1, np.array([[1., 0., -1.], [-1., .6, 1.]]), atol=0.01)
print("Exercise 5.1, Task 3 passed! ")
````

### Exercise 5.4: Histogram of array elements

#### Problem Description

The bin edges need to be equally spaced between the array minimum and maximum. A bin between the two edges $a$ and $b$ ($a \lt b$) should include all $v$ elements of the array where $a \leq v \lt b$, except for the last bin, where $a \leq v \leq b$ (with $b$ being the array maximum in this case). `n_bins` is assumed to be at least 1. `arr` is assumed to have at least two elements that are different.

The function should return two arrays:
- An integer array with the shape `(n_bins,)` containing the number of elements of the array falling into each bin.
- A float array with a shape of `(n_bins+1,)` containing the edges of the bins in ascending order. The first element should be equal to the array minimum and the last element equal to the array maximum.

**Constraint:** Do not rely on functions that directly produce a histogram (e.g., `np.histogram()`).

Example:
```
arr = np.array([0., 3., 3.5, 1.5, 2., 4.])
histogram(arr, n_bins=2) ->
    ndarray([2, 4]),
    ndarray([0., 2., 4.])
```

````python
# Implement your solution below








# Test cases
arr1 = np.array([0., 3., 3.5, 1.5, 2., 4.])
hist1, bin_edges1 = histogram(arr1, n_bins=2)
assert np.all(hist1 == [2, 4])
assert np.allclose(bin_edges1, [0., 2., 4.])

arr2 = np.array([[10., 11.], [5., 15.], [6., 14.]])
hist2, bin_edges2 = histogram(arr2, n_bins=4)
assert np.all(hist2 == [2, 0, 2, 2])
assert np.allclose(bin_edges2, [5., 7.5, 10., 12.5, 15.])

arr3 = np.array([[[-10.], [-8.]]])
hist3, bin_edges3 = histogram(arr3, n_bins=1)
assert np.all(hist3 == [2])
assert np.allclose(bin_edges3, [-10., -8.])

print("Exercise 5.4 passed! ")
````
