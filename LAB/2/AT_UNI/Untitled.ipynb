```python
def even_numbers(limite):
    for i in range(0, limite, 2):
        yield i
```

```python
generator = even_numbers(10)
```

```python
# DIFFERENCE BETWEEN str and repr
```

```python
class Playlist:
    def __init__(self, name):
        self.name = name
        self.tracks = []
    def __repr__(self):
        return f"Playlist({self.name!r}, {len(self.tracks)} tracks)"
    
    def __str__(self):
        return f"String: Playlist({self.name!r}, {len(self.tracks)} tracks)"
        
print(Playlist("Roadtrip"))
Playlist("Roadtrip")
```
