The number of categories is specified by the `n` parameter. The output is a 2-dimensional array of floating-point (float) type, where the slice `[i,:]` represents the one-hot encoding of the i-th element of the input array.


<font color='gray'>_In one-hot encoding, the `k` category labels (which are natural numbers) are converted into an `n`-dimensional vector, where `n` is the number of categories. Every element in the vector is 0, except for the k-th element, which is 1. We use one-hot encoding for multi-class classification: for example, when we want to determine what is depicted in photographs. Let’s assume we have 3 categories: dog, cat, and parrot. Each photo depicts exactly one of these, so we can assign a category label to each photo. For example, 0 for dog, 1 for cat, and 2 for parrot. When classifying photos using machine learning methods, we typically do not estimate the category label directly, but rather the probability that a given photo depicts a dog, a cat, or a parrot. We use one-hot encoding to represent the ground truth (true) label vectors, which we use to train the model. For example, if our image contains a parrot, the corresponding true label vector is `[0., 0., 1.]`, since the image has a 0 probability of being a dog or a cat, but a 1 probability of being a parrot._</font>
