## Python vs. NumPy - Speed comparison

### The multiplication of two randomly generated matrices

#### Naive Python implementation

```python
# Pure Python implementation (matrices: lists of lists)

import random

def multiply_random_matrices_naive(matrix_dim):
  matrix1 = [[random.random() for col_idx in range(matrix_dim)] for row_idx in range(matrix_dim)]
  matrix2 = [[random.random() for col_idx in range(matrix_dim)] for row_idx in range(matrix_dim)]
  matrix_out = [[0. for col_idx in range(matrix_dim)] for row_idx in range(matrix_dim)]

  for row1_idx in range(matrix_dim):
    for col2_idx in range(matrix_dim):
      for item_idx in range(matrix_dim):
        matrix_out[row1_idx][col2_idx] += matrix1[row1_idx][item_idx] * matrix2[item_idx][col2_idx]

  return matrix_out
```

#### Numpy implementation

```python
# Numpy implementation (matrices: ndarrays)

import numpy as np

def multiply_random_matrices_npy(matrix_dim):
  matrix1 = np.random.rand(matrix_dim, matrix_dim)
  matrix2 = np.random.rand(matrix_dim, matrix_dim)

  return np.matmul(matrix1, matrix2)
```

#### Benchmark

```python
import time

# We measure the execution time

SIZE = 400

time_start = time.time()
multiply_random_matrices_naive(SIZE)
time_end = time.time()

print("Multiplying two ", SIZE, "x", SIZE, " random matrices in pure Python takes",\
          time_end - time_start, "seconds.")

time_start = time.time()
multiply_random_matrices_npy(SIZE)
time_end = time.time()

print("Multiplying two ", SIZE, "x", SIZE, " random matrices in Numpy takes",\
          time_end - time_start, "seconds.")
```

### Counting numbers that are divisible by either 3 or 5

#### Naive Python implementation

```python

def count_special_numbers_naive(n):
    my_list = list(range(n))

    c = 0
    for elem in my_list:
        if (elem % 3 == 0) or (elem % 5 == 0):
            c += 1

    return c
```

#### (Efficeint) Numpy implementation

```python

def count_special_numbers_npy(n):
    my_array = np.arange(n, dtype=np.int32)
    return np.count_nonzero((my_array % 3 == 0) | (my_array % 5 == 0))
```

#### Benchmark

```python
# We measure the execution time

LEN = 100_000_000

time_start = time.time()
count_special_numbers_naive(LEN)
time_end = time.time()

print(f"Counting numbers divisible by 3 or 5 in the range [0, {LEN}) in pure Python takes {time_end - time_start} seconds.")

time_start = time.time()
count_special_numbers_npy(LEN)
time_end = time.time()

print(f"Counting numbers divisible by 3 or 5 in the range [0, {LEN}) in NumPy takes {time_end - time_start} seconds.")
```
