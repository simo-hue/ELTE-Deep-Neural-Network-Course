## A worked row

One row describes a motor. For fault classification, `fault` is the answer, `motor_id` identifies the motor, and `temp`, `rpm`, and `vibration` are measurements available at prediction time. For row `m02`, write down the three input values and the class ID. Look for them again after indexing the arrays.

````python
from pathlib import Path
from tempfile import TemporaryDirectory
import numpy as np

practice_dir = TemporaryDirectory()
folder = Path(practice_dir.name)
raw_file = folder / "motors.txt"
raw_file.write_text("""motor_id;temp;fault;rpm;vibration
m01;20;0;1000;0.2
m02;25;1;1200;0.8
m03;;0;1100;0.3
m04;80;1;?;0.9
m05;30;;1300;0.5
m06;28;0;1150;NA
""", encoding="utf-8")
header, *lines = raw_file.read_text(encoding="utf-8").splitlines()
names = header.split(";")
rows = [line.split(";") for line in lines]
assert all(len(row) == len(names) for row in rows)
print(dict(zip(names, rows[1])))
````

A missing **field** still occupies a column. Splitting `m03;;0;1100;0.3` on `;` preserves the empty temperature field. A missing **file** raises `FileNotFoundError` when Python tries to open it. Here we already have the file, so only missing numeric fields become `np.nan`.

````python
def number(token):
    return np.nan if token.strip() in {"", "?", "NA"} else float(token)

# Keep IDs separately. The numeric table inherits the remaining header order.
ids = np.array([row[0] for row in rows])
columns = names[1:]
table = np.array([[number(token) for token in row[1:]] for row in rows])
print(columns)
print(table)
````

## The shape decision

`:` selects all entries on an existing axis. `None` inserts an axis of length one. An integer index selects one position and removes that axis. Start with the target shape your loss expects **after batching**, then work backwards to one sample.

| Training contract | Stored `y` | `y[i]` | Batch of B labels |
|---|---|---|---|
| One class index per example (cross-entropy with class indices) | `(N,)` integer | `()` | `(B,)` |
| One numeric output per example, model outputs `(B, 1)` for regression | `(N, 1)` float | `(1,)` | `(B, 1)` |
| Two numeric outputs per example | `(N, 2)` float | `(2,)` | `(B, 2)` |

A scalar has shape `()`. A length-one vector has shape `(1,)`. A single-column matrix has shape `(N, 1)`. A regression model that outputs `(B,)` can also use flat targets. Match the prediction and target shapes for elementwise regression loss.
