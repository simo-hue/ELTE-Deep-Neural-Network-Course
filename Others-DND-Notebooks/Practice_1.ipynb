## Functions and return values

Functions are defined with `def`. A function can return multiple values; Python packs them into a tuple, which can then be unpacked into separate variables.

The last example also illustrates notebook/function scope: rebinding the local parameter `v` does not rebind the variable `a` in the calling scope.

```python
def add_one(value):
    return value, value + 1
```

```python
a, e = add_one(2)
```

## Reading and writing files

`open()` gives access to files. Using a `with` block is preferred because the file is closed automatically when the block ends.

The reading examples expect a `test.txt` file in the current working directory. The writing example creates or overwrites `test2.txt`.

```python
with open('test.txt', 'w') as f:
    f.write('first line\nsecond line\n')

f = open('test.txt', 'r')
f.readlines()
f.close()
```

```python
with open('test.txt') as f:
    print(f.readlines())
```

```python
with open('test2.txt', 'w') as w:
    w.write('abc')
```

## `map`, `filter`, `lambda`, and `zip`

These tools support compact transformations and combinations of iterables:

- `map(function, iterable)` applies a function to every element.
- `filter(function, iterable)` keeps elements for which the function returns `True`.
- `lambda` creates a small anonymous function.
- `zip(...)` combines iterables element by element.

The following cells show the examples demonstrated during the practice.

```python
list(map(lambda x: x + 1, [1, 2, 3]))
```

```python
l = list(range(10))
l
```

```python
list(filter(lambda x: x % 2 == 0, l))
```

```python
dict(zip(['a', 'b', 'c'], [1, 2, 3]))
```

## Indexing and slicing

Slices use the form `start:stop:step`. Omitting values is allowed; for example, `::2` selects every second element.

The last cell intentionally indexes past the end of a list to demonstrate an `IndexError`.

```python
"abcdef"[::2]
```

```python
try:
    [1, 2, 3, 4][5]
except IndexError as e:
    print(f'{type(e).__name__}: {e}')
```
